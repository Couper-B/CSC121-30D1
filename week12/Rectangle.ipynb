```python
# NAME: rectangle.ipynb
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# 1. Define Calculate_Rectangle(L,W)
#  - Accept L/Length and W/Width
def Calculate_Rectangle(L,W):
#  - Calculate A/Area(L*W)
  A = L * W
#  - Calculate P/Perimeter(2*(L+W))
  P = 2 * (L + W)
#  - Return both results as a single tuple
  T = (A,P)
  return (T)

# PROGRAM LOGIC
# 1. Request Length & Width from user(convert to flt)
L = 6
W = 11
# 2. Call Calculate_Rectangle(L,W)
Calculate_Rectangle(L,W)
# 3. Unpack the returned Tuple into two seperate variables called My_Area & My_Perimeter
# 4. Print Result
print(Calculate_Rectangle(L,W))
```
