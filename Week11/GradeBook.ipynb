```python
# NAME: GradeBook.ipynb
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# 1. Create Add_Students(Grades)
def Add_Students(Grades):
  Name = "name"
#  - Use while loop till "done" is entered
  while Name != "done":
#  - Collect Student's name
    Name = str(input('Enter student name (or "done"): '))
#  - Use name as key then collect their grade to use as value
    if Name != "done":
      Grades[Name] = float(input('Enter '+Name+"'s grade: "))

# 2. Create Calculate_Average(Grades)
#  - Collect Values from "Grades" calculate and return average

# PROGRAM LOGIC
# Create empty dictionary called "Grades".
Grades = {}
# Call Add_Students to fill "Grades".
Add_Students(Grades)
# Call Calculate_Average and display class average.
# Print final report showing all students and their grades.
print(Grades)
```
