```python
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# 1. Define Is_Palindrome():
def Is_Palindrome():
#  -accepts one parameter(the word to be checked)
#  -Not case sensitive
  Parameter = Input
  Parameter = Parameter.lower()
  length = len(Parameter)
  Digit = 0
#  -returns True or False depending on word entered
  while length>0:
    if Parameter[Digit] == Parameter[length-1]:
      Digit = Digit + 1
      length = length-1
      return True
    else:
     break

# PROGRAM LOGIC
# 1. Prompt user for word
Input = input("Please enter a word: ").lower()
# 2. Call Is_Palindrome()
Is_Palindrome()
# 3. Print a message indicating result
if Is_Palindrome() == True:
  print("That is a palindrome!")
else:
  print("That is not a Palindrome.")
```
