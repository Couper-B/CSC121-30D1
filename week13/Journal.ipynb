```python
# NAME: journal.ipynb
# IMPORT STATEMENTS
import os

# FUNCTION DEFINITIONS
# 1. Define Write_entry
def Write_entry(filename='Journal.txt'):
#  -It must open the file in append mode ('A').
  Append = open(filename, "a")
#  -It should write the entry to the file, ensuring each new entry starts on a new line.
  Entry = str(input("What would you like to write? "))
  Append.write(Entry + "\n")
  Append.close()
# 2. Define Read_journal
def Read_journal(filename='Journal.txt'):
#  -It should open the file in read mode ('R').
  Read = open(filename, "r")
#  -It should read the entire contents and print them to the screen so the user can see their past entries.
  print('---Your Journal---'+'\n',Read.read())
  Read.close()
# PROGRAM LOGIC
# Wrap this in a loop so the user can keep using the app until they type "quit".
while True:
# Ask the user if they want to (W)rite a new entry or (R)ead their journal.
  Mode = input("Do you want to (W)rite a new entry or (R)ead your journal? ").upper()
# If they choose 'W', ask for their thoughts and call write_entry.
  if Mode == 'W':
    Write_entry('Journal.txt')
# If they choose 'R', call read_journal.
  elif Mode == 'R':
    if not os.path.exists('Journal.txt'):
      print("No entries yet.")
    else:
      Read_journal('Journal.txt')
# If they type 'quit', exit the loop.
  elif Mode == 'QUIT':
    break
```
