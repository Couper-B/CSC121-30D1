```python
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# 1. make function Add_Players
def Add_Players(Roster):
  Name = ""
  #  - Uses a while loop
  while Name != "quit":
    #  - accepts player name from user
    #  - that adds player names to list until "quit" is entered
    Name = input('Enter player name (or "quit" to stop): ')
    if Name != "quit":
      Roster.append(Name)
  return Roster
# 2. make function Is_On_Roster
#  - accepts list and can check for a specific name
def Is_On_Roster(Roster, Name):
#  - should return True if name is on the list and False if it's not
  if Name in Roster:
    return True
  else:
    return False


# PROGRAM LOGIC
# 1. create empty list
Roster = []
# 2. call Add_Players
Add_Players(Roster)
# 3. request name to search for
Name = input("Enter name to search for: ")
# 4. call Is_On_Roster and use returned value to print message in response
Is_On_Roster(Roster, Name)
if Is_On_Roster(Roster, Name) == True:
  print(Name, "is on the roster.")
else:
  print(Name, "is not on the roster.")
```
